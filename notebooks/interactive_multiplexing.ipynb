# Interactive Multiplexing Explorer

This notebook provides a small UI for composing multiplexing figures from the public supplementary-material dataset. Each enabled layer can choose its own data source, visualization, field, and color style.

## Controls

- `Global Controls`: shared settings such as epoch, glyph size, grid resolution, and colorbar visibility
- `Layer 1 / 2 / 3`: each layer can be enabled independently
- `Source`: choose from `testing`, `training`, or `training_loss`
- `Vis`: choose a compatible visualization for the selected source; `contour` now means line contours only, and `zebra_map` is a dedicated layered view with adjustable iso-level masking
- `Field`: choose the metric provided by the selected source
- `Colors`, `Alpha`, `Levels`, `Point`: layer-specific styling controls; `Colors` now selects a base colormap, and `zebra_map` uses `Levels` to derive zebra masking dynamically
- `Range`: switch between automatic range and manual `vmin/vmax`
- `Ticks`: choose automatic ticks, level-aligned ticks, fixed tick counts, or manual tick values
- Contextual controls: the UI now hides controls that are not relevant in the current mode, so manual-only inputs only appear when needed
